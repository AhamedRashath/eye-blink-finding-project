```python
import cv2
import numpy as np
import mediapipe as mp
from mediapipe.tasks import python
from mediapipe.tasks.python import vision

video_path = "/content/tranding video blinking eye #devranijethani #funnyvideo #comedyvideo #familyvlog #happy #shots #love.mp4"

cap = cv2.VideoCapture(video_path)

if not cap.isOpened():
    print("Error:File can't be opened")
else:
    print("successfully opened")
base_options = python.BaseOptions(model_asset_path='face_landmarker_v2_with_blendshapes.task')
options = vision.FaceLandmarkerOptions(base_options=base_options,
                                       output_face_blendshapes=True,
                                       output_facial_transformation_matrixes=True,
                                       num_faces=1)
detector = vision.FaceLandmarker.create_from_options(options)
blink_count = 0
counter = 0
EYE_AR_THRESH = 0.23
EYE_AR_CONSEC_FRAMES = 2

LEFT_EYE = [362, 385, 387, 263, 373, 380]
RIGHT_EYE = [33, 160, 158, 133, 153, 144]

def calculate_ear(eye_points, landmarks):
    # Extract x, y from our landmark tuples
    pt1 = np.array([landmarks[eye_points[1]][0], landmarks[eye_points[1]][1]])
    pt5 = np.array([landmarks[eye_points[5]][0], landmarks[eye_points[5]][1]])
    pt2 = np.array([landmarks[eye_points[2]][0], landmarks[eye_points[2]][1]])
    pt4 = np.array([landmarks[eye_points[4]][0], landmarks[eye_points[4]][1]])
    pt0 = np.array([landmarks[eye_points[0]][0], landmarks[eye_points[0]][1]])
    pt3 = np.array([landmarks[eye_points[3]][0], landmarks[eye_points[3]][1]])

    A = np.linalg.norm(pt1 - pt5)
    B = np.linalg.norm(pt2 - pt4)
    C = np.linalg.norm(pt0 - pt3)
    ear = (A + B) / (2.0 * C)
    return ear

while cap.isOpened():
    success, frame = cap.read()
    if not success:
        break

    rgb_frame = cv2.cvtColor(frame, cv2.COLOR_BGR2RGB)
    mp_image = mp.Image(image_format=mp.ImageFormat.SRGB, data=rgb_frame)

    detection_result = detector.detect(mp_image)

    if detection_result.face_landmarks:
        for face_landmarks in detection_result.face_landmarks:
            h, w, _ = frame.shape

            landmarks = [(int(pt.x * w), int(pt.y * h)) for pt in face_landmarks]
            left_ear = calculate_ear(LEFT_EYE, landmarks)
            right_ear = calculate_ear(RIGHT_EYE, landmarks)
            avg_ear = (left_ear + right_ear) / 2.0

            if avg_ear < EYE_AR_THRESH:
                counter += 1
            else:
                if counter >= EYE_AR_CONSEC_FRAMES:
                    blink_count += 1
                counter = 0

cap.release()
print(f"🎉 Total number of blinks: {blink_count}")
```
